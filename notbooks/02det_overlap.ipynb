```python
import sympy as sp
from sympy import *
from sympy.matrices import Matrix
import matplotlib.pyplot as plt
```

```python
for dim in range(3,6):
    s = sp.Symbol('s')

    g = [[1 if _d==d else s for _d in range(dim)] for d in range(dim)]

    G = Matrix(g)
    print(G.det())

    plot(G.det(), (s, 0, 1), title=f'dim {dim}', plot_backends=plt)

```
