```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np

# 🌟 1. 백엔드를 'pgf'로 강제 지정 (plt를 부르기 전에 설정해야 함!)
mpl.use("pgf")

# 🌟 2. PGF 전용 논문 스타일 환경 설정 (rcParams)
mpl.rcParams.update({+
    "font.family": "serif",               # 세리프체 사용
    "mathtext.fontset": "cm",             # 수식 폰트를 Computer Modern으로 지정
    "font.serif": ["Computer Modern Roman", "Times New Roman", "DejaVu Serif"],
    "axes.formatter.use_mathtext": True,  # 축의 숫자들도 수식 폰트로 통일
    "font.size": 12,
    "lines.linewidth": 2
})

fig, ax = plt.subplots(figsize=(8, 6)) # 🌟 크기를 인치(inch) 단위로 신중하게 설정 (논문 단락 너비에 맞게)

# 가상의 데이터 (영준님의 VQSD Fixed Rate 결과라고 가정)
x = np.linspace(0, 1, 100)
y_succ = 1 - 0.5 * x**2
y_fail = 0.5 * x**2

# 그래프 그리기 (LaTeX 수식을 마음껏 쓸 수 있습니다!)
ax.plot(x, y_succ, label=r"Success $P_{\text{succ}}(\lambda)$", color="dodgerblue")
ax.plot(x, y_fail, label=r"Failure $P_{\text{fail}}(\lambda)$", color="limegreen")

ax.set_xlabel(r"Fixed Rate ($\gamma$)")
ax.set_ylabel(r"Probability ($P$)")
ax.set_title(r"VQSD Result [COBYLA] ($\dim=4$, $\lambda=5.0$)")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()

# 🌟 3. 파일 저장하기
# 방법 A: .pgf 로 저장 (LaTeX 코드로 저장됨. 본문에 직접 삽입용)
plt.savefig("vqsd_result.pgf")
```
