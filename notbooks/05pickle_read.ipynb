```python
import pandas as pd
import numpy as np
import pickle

# 1. 에러를 우회하기 위한 '빈 껍데기' DATA 클래스를 선언합니다.
class DATA:
    pass

select = "2026_1_19_16_32__45"
# 2. 파일을 다시 불러옵니다. (이제 에러가 나지 않습니다!)
with open(f"data/pickle_SP/{select}", "rb") as f:
    data = pickle.load(f)

# 3. 빈 껍데기에 채워진 실제 데이터(속성들)를 딕셔너리 형태로 뽑아봅니다.
extracted_dict = vars(data) # data.__dict__ 와 동일합니다.

# 1. 클래스 의존성이 제거된 '안전한' 순수 딕셔너리 형태로 피클 다시 저장하기
# 나중에 불러올 때 더 이상 DATA 클래스가 없어도 에러가 나지 않습니다.
safe_pickle_path = f"data/pickle_SP/{select}_fixed.pkl"
with open(safe_pickle_path, "wb") as f:
    pickle.dump(extracted_dict, f)
print(f"✅ 안전한 피클 파일이 저장되었습니다: {safe_pickle_path}")

# 2. 2D Crosstalk 배열만 따로 빼내서 CSV로 저장하기 (분석용)
crosstalk_2d = extracted_dict.get('Crosstalk_array_2d')
if crosstalk_2d is not None:
    df_crosstalk = pd.DataFrame(crosstalk_2d)
    crosstalk_csv_path = "Crosstalk_array_2d.csv"
    df_crosstalk.to_csv(crosstalk_csv_path, index=False, header=False)
    print(f"✅ 2D 누화 행렬이 저장되었습니다: {crosstalk_csv_path}")

# 3. 배열(Array)과 리스트를 제외한 '단일 설정값(메타데이터)'들만 모아서 CSV로 저장
metadata = {
    k: v for k, v in extracted_dict.items()
    if not isinstance(v, (np.ndarray, list, tuple))
}
df_meta = pd.DataFrame([metadata])
meta_csv_path = "experiment_metadata.csv"
df_meta.to_csv(meta_csv_path, index=False)
print(f"✅ 실험 설정값(메타데이터)이 저장되었습니다: {meta_csv_path}")
```
